Ejercicio 1

```python
import numpy as np

n = 10
```

```python
def factorial(n):
    resultado = 1
    
    for i in range(1, n + 1):
        resultado = resultado * i
    
    return resultado

factorial(n)
```

```python
V1 = np.random.randint(1, 11, n)
V2 = np.random.randint(1, 11, n)

print(V1)
print(V2)
```

```python
V3 = []

for i in range(n):
    if V1[i] < V2[i]:
        V3.append(0)
    else:
        V3.append(1)

V3 = np.array(V3)

print(V3)
```

```python
V1_col = V1.reshape(n, 1)
V2_col = V2.reshape(n, 1)
V3_col = V3.reshape(n, 1)

matriz = np.concatenate((V1_col, V2_col, V3_col), axis=1)

print(matriz)
```

Ejercicio 2

```python
m = 4
n = 3
```

```python
A = np.zeros((m, n), dtype=int)

print(A)
```

```python
j = 0

while j < n:
    A[:, j] = np.random.randint(-1, 4, m)
    j = j + 1

print(A)

A_original = A.copy()
```

```python
A_while = A_original.copy()

i = 0

while i < m:
    j = 0
    
    while j < n:
        if A_while[i, j] < 0:
            A_while[i, j] = -1
        else:
            A_while[i, j] = 5
        
        j = j + 1
    
    i = i + 1

print(A_while)
```

```python
A_for = A_original.copy()

for i in range(m):
    for j in range(n):
        if A_for[i, j] < 0:
            A_for[i, j] = -1
        else:
            A_for[i, j] = 5

print(A_for)
```

```python
np.array_equal(A_while, A_for)
```
